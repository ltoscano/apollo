```
import json, sys
from pathlib import Path
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root / 'src'))
from qrand import report
full, power = report.load_controls('full'), report.load_controls('power')
print('controlli 10^8:', sorted(full)); print('repliche 10^7:', {k: len(v) for k, v in power.items()})
```

```
from IPython.display import Markdown
Markdown(report.full_controls_table(full))
```
